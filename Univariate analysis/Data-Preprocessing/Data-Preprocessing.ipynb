```python
import pandas as pd
dataset = pd.read_csv("Placement.csv")
```

```python
dataset.isnull().sum()
```

```python
dataset.isna().sum()
```

```python
dataset["salary"]=dataset["salary"].fillna(0)
```

```python
dataset.isnull().sum()
```

```python
def quanQual(dataset):
    quan=[]
    qual=[]
    for columnName in dataset.columns:
        if(dataset[columnName].dtype=="str"):
            qual.append(columnName)
        else:
            quan.append(columnName)
    return quan,qual        
            
```

```python
quan,qual = quanQual(dataset)
```

```python
dataset[quan]
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp_mean = SimpleImputer(missing_values=np.nan, strategy='mean')
imp_mean.fit(dataset[quan])
df=imp_mean.transform(dataset[quan])
```

```python
df
```

```python
df = pd.DataFrame(df, columns=quan)
```

```python
cate = dataset[qual]
```

```python
cate
```

```python
two = [df,cate]
```

```python
preprocessed = pd.concat(two,axis=1)
```

```python
preprocessed
```

```python
preprocessed.isnull().sum()
```

```python
preprocessed.to_csv("Preplacementdata.csv",index=False)
```
